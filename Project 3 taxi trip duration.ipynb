## Problem Statement

New York City is one of the busiest cities in the world, and thousands of taxi trips take place every day. The time required to complete a taxi trip can vary greatly, even when the distance between two locations is almost the same. Traffic, time of day, day of the week, pickup and drop-off locations, and other factors can all affect how long a trip takes.

In this project, I am working with a large dataset of NYC taxi trips to understand the factors that influence trip duration. Instead of focusing only on building a highly complex prediction model, the main goal is to explore the data, identify useful patterns, and create meaningful features from the available information.

By analyzing factors such as trip distance, pickup and drop-off locations, time, day, and travel direction, this project aims to understand why some taxi trips take longer than others. The analysis can help provide useful insights into taxi travel patterns and the factors that have the greatest impact on trip duration.

The overall goal of this project is to turn raw taxi trip data into meaningful information that can help us better understand and predict travel time in New York City.

```python
import pandas as pd
import plotly.express as px

# Convert pickup_datetime to datetime
df['pickup_datetime'] = pd.to_datetime(df['pickup_datetime'])

# Extract date and hour
df['pickup_date'] = df['pickup_datetime'].dt.date
df['pickup_hour'] = df['pickup_datetime'].dt.hour

# Count trips by date and hour
trip_counts = (
    df.groupby(['pickup_date', 'pickup_hour'])
      .size()
      .reset_index(name='trips')
)

# Animated plot
fig = px.bar(
    trip_counts,
    x='pickup_hour',
    y='trips',
    animation_frame='pickup_date',
    range_x=[0, 23],
    range_y=[0, trip_counts['trips'].max()],
    title='Animated Taxi Trips Over Time',
    labels={
        'pickup_hour': 'Hour of Day',
        'trips': 'Number of Trips',
        'pickup_date': 'Date'
    }
)

fig.show()
```

```python
import pandas as pd
import plotly.express as px

# Convert pickup_datetime to datetime
df['pickup_datetime'] = pd.to_datetime(df['pickup_datetime'])

# Passenger category
df['passenger_category'] = pd.cut(
    df['passenger_count'],
    bins=[0, 1, 2, 4, 6],
    labels=['Solo', 'Small Group', 'Medium Group', 'Large Group']
)

# Time category
df['hour'] = df['pickup_datetime'].dt.hour

df['time_category'] = pd.cut(
    df['hour'],
    bins=[-1, 5, 11, 17, 23],
    labels=['Night', 'Morning', 'Afternoon', 'Evening']
)

# Calculate trip distance using latitude and longitude
from math import radians, sin, cos, sqrt, atan2

def calculate_distance(row):
    lat1 = radians(row['pickup_latitude'])
    lon1 = radians(row['pickup_longitude'])
    lat2 = radians(row['dropoff_latitude'])
    lon2 = radians(row['dropoff_longitude'])

    dlat = lat2 - lat1
    dlon = lon2 - lon1

    a = sin(dlat/2)**2 + cos(lat1) * cos(lat2) * sin(dlon/2)**2
    c = 2 * atan2(sqrt(a), sqrt(1-a))

    return 6371 * c

df['distance_km'] = df.apply(calculate_distance, axis=1)

# Distance category
df['distance_category'] = pd.cut(
    df['distance_km'],
    bins=[0, 2, 5, 10, float('inf')],
    labels=['Short', 'Medium', 'Long', 'Very Long']
)

# Remove missing categories
sunburst_df = df.dropna(
    subset=['passenger_category', 'time_category', 'distance_category']
)

# Create Sunburst plot
fig = px.sunburst(
    sunburst_df,
    path=[
        'passenger_category',
        'time_category',
        'distance_category'
    ],
    title='Taxi Trips: Passenger → Time → Distance Category'
)

fig.show()
```

```python
import pandas as pd
import plotly.express as px
import imageio.v2 as imageio
import os

# Convert pickup_datetime to datetime
df['pickup_datetime'] = pd.to_datetime(df['pickup_datetime'])

# Extract date and hour
df['pickup_date'] = df['pickup_datetime'].dt.date
df['pickup_hour'] = df['pickup_datetime'].dt.hour

# Count trips by date and hour
trip_counts = (
    df.groupby(['pickup_date', 'pickup_hour'])
      .size()
      .reset_index(name='trips')
)

# Create folder for frames
os.makedirs('gif_frames', exist_ok=True)

# Store frame file names
frames = []

# Create one image for each date
for i, date in enumerate(sorted(trip_counts['pickup_date'].unique())):

    data = trip_counts[trip_counts['pickup_date'] <= date]

    fig = px.bar(
        data,
        x='pickup_hour',
        y='trips',
        title=f'Taxi Trips Over Time - {date}',
        labels={
            'pickup_hour': 'Hour of Day',
            'trips': 'Number of Trips'
        }
    )

    fig.update_layout(
        xaxis=dict(dtick=1),
        yaxis=dict(range=[0, trip_counts['trips'].max() * 1.1])
    )

    frame_path = f'gif_frames/frame_{i:03d}.png'

    fig.write_image(frame_path)
    frames.append(frame_path)

# Create GIF
with imageio.get_writer(
    'taxi_trips_animation.gif',
    mode='I',
    duration=0.5
) as writer:

    for frame in frames:
        writer.append_data(imageio.imread(frame))

print("GIF saved successfully as taxi_trips_animation.gif")
```
