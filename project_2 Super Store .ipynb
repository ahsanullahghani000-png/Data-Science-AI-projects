# **4** **Feature Engineering**

***Create Profit Ratio = Profit/ Sale***

```python
# New columns: Profit Ratio
df['Profit Ratio'] = df['Profit'] / df['Sales']

print(df[['Sales', 'Profit', 'Profit Ratio']].head(30))
```

***Extract Year and Month From Order Dates***

```python

#conver order data to date format
df['Order Date'] = pd.to_datetime(df['Order Date'])
df['Order Year'] = df['Order Date'].dt.year
df['Order Month'] = df['Order Date'].dt.month
print(df[['Order Date', 'Order Year', 'Order Month']])
```

***Identify Top performing and loaa making area***

```python
# Group by Category and sum the Profit for each category
category_profit = df.groupby('Category')['Profit'].sum()

# Sort values from lowest to highest
category_profit = category_profit.sort_values()

print("Category wise Profit:")
print(category_profit)

# Group by Sub-Category and sum the Profit for each sub-category
subcat_profit = df.groupby('Sub-Category')['Profit'].sum()

# Sort values from lowest to highest
subcat_profit = subcat_profit.sort_values()

print("\nTop 5 Performing Sub-Categories:")
print(subcat_profit.tail(5))   # last 5 = highest profit

print("\nLoss Making Sub-Categories:")
print(subcat_profit.head(5))   # first 5 = lowest / negative profit
```

# **5** **Advance Analysis**

***Identify Loss Making Product***

```python

# Group by Product Name and sum the Profit for each product
product_profit = df.groupby('Product Name')['Profit'].sum()

# Sort values from lowest to highest
product_profit = product_profit.sort_values()

# Filter only products with negative profit (loss making)
loss_products = product_profit[product_profit < 0]

print("Total Loss Making Products:", len(loss_products))
print("\nTop 10 Loss Making Products:")
print(loss_products.head(10))
```

***Region Wise Proftability Analysis***

```python
# Group by Region and calculate total Sales, Profit and Quantity
region_analysis = df.groupby('Region')[['Sales', 'Profit', 'Quantity']].sum()

# Calculate Profit Ratio (Profit / Sales) for each region
region_analysis['Profit Ratio'] = region_analysis['Profit'] / region_analysis['Sales']

# Sort by Profit from highest to lowest
region_analysis = region_analysis.sort_values('Profit', ascending=False)

print(region_analysis)
```

***Montly And Yearly Trend***

```python
# Convert Order Date to proper date format
df['Order Date'] = pd.to_datetime(df['Order Date'])

# Extract Year and Month
df['Order Year'] = df['Order Date'].dt.year
df['Order Month'] = df['Order Date'].dt.month

# Yearly Trend
yearly_trend = df.groupby('Order Year')[['Sales', 'Profit']].sum()
print("Yearly Trend:")
print(yearly_trend)

# Monthly Trend (combined across all years)
monthly_trend = df.groupby('Order Month')[['Sales', 'Profit']].sum()
print("\nMonthly Trend:")
print(monthly_trend)
```

***Discount VS Profit Relationship***

```python
# Check correlation between Discount and Profit
correlation = df['Discount'].corr(df['Profit'])
print("Correlation between Discount and Profit:", correlation)

# Group by Discount value and check average Profit
discount_profit = df.groupby('Discount')['Profit'].mean()
print("\nAverage Profit by Discount Level:")
print(discount_profit)
```

# **6** **Visulization**

***Sale Trend Over time***

```python
import pandas as pd
import plotly.express as px
# Convert Order Date to proper date format
df['Order Date'] = pd.to_datetime(df['Order Date'])

# Extract Year-Month for grouping
df['Year Month'] = df['Order Date'].dt.to_period('M').astype(str)

# Group by Year Month and calculate total Sales
sales_trend = df.groupby('Year Month')['Sales'].sum().reset_index()

# Create Line Chart
fig = px.line(sales_trend, x='Year Month', y='Sales', title='Sales Trend Over Time')
fig.show()
```

***Profit By Catagory***

```python
# Group by Category and calculate total Profit
category_profit = df.groupby('Category')['Profit'].sum().reset_index()

# Create Bar Chart
fig = px.bar(category_profit, x='Category', y='Profit', title='Profit By Category')
fig.show()
```

***Correlation Heatmap***

```python
# Select numeric columns only
numeric_df = df[['Sales', 'Profit', 'Quantity', 'Discount']]

# Calculate correlation between columns
corr_matrix = numeric_df.corr()

# Create Heatmap
fig = px.imshow(corr_matrix, text_auto=True, color_continuous_scale='RdBu', title='Correlation Heatmap')
fig.show()
```

***Region Comparison chart***

```python

# Group by Region and calculate total Sales and Profit
region_data = df.groupby('Region')[['Sales', 'Profit']].sum().reset_index()

# Create grouped Bar Chart
fig = px.bar(region_data, x='Region', y=['Sales', 'Profit'], barmode='group', title='Region Comparison - Sales vs Profit')
fig.show()
```

## Things I found out

- Technology makes the most profit, but Furniture is barely profitable even though its sales are almost as high as Technology
- Tables and Bookcases are actually losing money for the store
- West region is doing the best overall, Central is the weakest in terms of profit

- Sales spike every year in November-December, which makes sense because of holiday shopping
