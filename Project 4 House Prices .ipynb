```python
import pandas as pd 
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats
```

```python
# 1) PRICE CHANGE % - how much price changed compared to the previous row
df['price_pct_change'] = df['price'].pct_change() * 100

# 2) RATIO - price for each square foot
df['price_per_sqft'] = df['price'] / df['sqft_living']

# 3) RATIO - average price for each bedroom
df['price_per_bedroom'] = df['price'] / df['bedrooms']

print(df[['price', 'sqft_living', 'price_pct_change', 'price_per_sqft', 'price_per_bedroom']].head(10))
print("\nAverage price change %:", df['price_pct_change'].mean())
print("Average price per sqft:", df['price_per_sqft'].mean())
```

 **Work with fractions and proportions from data**

```python
# how many houses have more than 3 bedrooms - getting the fraction
fraction_more_than_3bed = (df['bedrooms'] > 3).sum() / len(df)

# how many houses have 2 or more bathrooms
fraction_2plus_bath = (df['bathrooms'] >= 2).sum() / len(df)

# bathroom to bedroom ratio for each house - gives an idea of house size
df['bath_bed_ratio'] = df['bathrooms'] / df['bedrooms']

# each house's sqft as a share of the total sqft (all shares add up to 100%)
df['sqft_share'] = df['sqft_living'] / df['sqft_living'].sum()

print(f"Houses with more than 3 bedrooms: {fraction_more_than_3bed:.2%}")
print(f"Houses with 2+ bathrooms: {fraction_2plus_bath:.2%}")
print(df[['bedrooms', 'bathrooms', 'bath_bed_ratio', 'sqft_share']].head(10))
```

# **2. ALGEBRA (EQUATIONS)**

</div>

```python
# our price model - price depends on sqft and bedrooms, plus a base amount
# Price = a*(sqft) + b*(bedrooms) + c
equation_str = "Price = a*sqft_living + b*bedrooms + c"
print(equation_str)

# a, b, c are unknowns - we don't know their values yet
# a = how much price increases per extra sqft
# b = how much price increases per extra bedroom
# c = base price (like a starting price even with 0 sqft, just theoretically)
print("a = price per sqft")
print("b = price per bedroom")
print("c = base price / intercept")
```

```python
# building X matrix (features) and y vector (target price)
X = df[['sqft_living', 'bedrooms']].to_numpy(dtype=float)
X = np.column_stack([X, np.ones(len(X))])   # adding a column of 1s for c
y = df['price'].to_numpy(dtype=float)

# solving for a, b, c using least squares (best fit across ALL rows, not just 3)
coeffs, residuals, rank, sv = np.linalg.lstsq(X, y, rcond=None)
a, b, c = coeffs

print(f"a (price per sqft):     {a:.2f}")
print(f"b (price per bedroom):  {b:.2f}")
print(f"c (base price):         {c:.2f}")

print(f"\nFinal equation: Price = {a:.2f}*sqft_living + {b:.2f}*bedrooms + {c:.2f}")
```

# **5. SOLVE SYSTEM OF EQUATIONS**

</div>

```python
# taking first 3 houses from dataset
sample = df.iloc[:3]

# building matrix A (sqft, bedrooms, and 1)
A = np.array([
    [sample['sqft_living'].iloc[0], sample['bedrooms'].iloc[0], 1],
    [sample['sqft_living'].iloc[1], sample['bedrooms'].iloc[1], 1],
    [sample['sqft_living'].iloc[2], sample['bedrooms'].iloc[2], 1]
])

# putting prices into B
B = np.array([
    sample['price'].iloc[0],
    sample['price'].iloc[1],
    sample['price'].iloc[2]
])

print("A:\n", A)
print("B:", B)

# METHOD 1: Inverse method
A_inverse = np.linalg.inv(A)
answer_1 = A_inverse @ B
print("\nMethod 1 (Inverse) - a, b, c:", answer_1)

# METHOD 2: NumPy's solve function
answer_2 = np.linalg.solve(A, B)
print("Method 2 (NumPy solve) - a, b, c:", answer_2)
```

# **6. ADVANCED LINEAR ALGEBRA**

</div>

```python
# taking 3 houses, 3 features each (sqft, bedrooms, bathrooms)
M = df[['sqft_living', 'bedrooms', 'bathrooms']].to_numpy(dtype=float)[:3]

print("Matrix M:\n", M)

# DOT PRODUCT - combining two rows into a single number
dot_result = np.dot(M[0], M[1])
print("\nDot product of row 0 and row 1:", dot_result)

# MATRIX MULTIPLICATION
matmul_result = M @ M.T
print("\nMatrix multiplication (M @ M.T):\n", matmul_result)

# DETERMINANT - a single number that tells us if the matrix can be inverted
det_result = np.linalg.det(M)
print("\nDeterminant of M:", det_result)

# INVERSE - only works if determinant is not zero
if abs(det_result) > 1e-6:
    inverse_result = np.linalg.inv(M)
    print("\nInverse of M:\n", inverse_result)
else:
    print("\nMatrix can't be inverted (determinant is 0)")
```

# **7. EIGENVALUES & EIGENVECTORS**


```python
# taking 3 features from dataset
features = df[['sqft_living', 'bedrooms', 'bathrooms']].to_numpy(dtype=float)

# covariance matrix - shows how features vary together
cov_matrix = np.cov(features, rowvar=False)
print("Covariance matrix:\n", cov_matrix)

# calculating eigenvalues and eigenvectors
eigenvalues, eigenvectors = np.linalg.eig(cov_matrix)

print("\nEigenvalues:", eigenvalues)
print("\nEigenvectors:\n", eigenvectors)

# how much each eigenvalue contributes (as a percentage)
explained = eigenvalues / eigenvalues.sum()
print("\nVariance explained by each:", explained)
```

# **8. SVD (SINGULAR VALUE DECOMPOSITION)**

</div>

```python
# taking 3 features from dataset
features = df[['sqft_living', 'bedrooms', 'bathrooms']].to_numpy(dtype=float)

# applying SVD
U, S, Vt = np.linalg.svd(features, full_matrices=False)

print("U shape:", U.shape)
print("Singular values (S):", S)
print("Vt shape:", Vt.shape)

# checking we can rebuild the original matrix from U, S, Vt
rebuilt = U @ np.diag(S) @ Vt
print("\nRebuilt matrix matches original:", np.allclose(features, rebuilt))

# how much of the data's pattern is captured by the top value
top_value_importance = (S[0]**2) / np.sum(S**2)
print(f"\nTop singular value captures {top_value_importance:.1%} of the data's pattern")
```
