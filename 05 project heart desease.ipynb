```python
# Numerical columns 
numerical_cols = ["age", "trestbps", "chol", "thalach", "oldpeak"]

# Mean: average value
mean_values = df[numerical_cols].mean()

# Median: middle value
median_values = df[numerical_cols].median()

# Mode: most frequent value (mode() can return more than one value, so we take the first row)
mode_values = df[numerical_cols].mode().iloc[0]

central_tendency = pd.DataFrame({
    "Mean": mean_values,
    "Median": median_values,
    "Mode": mode_values
}).round(2)

print("Measures of Central Tendency:")
display(central_tendency)
```

```python
# Variance: how spread out the values are (squared units)
variance_values = df[numerical_cols].var()

# Standard deviation: spread in the original units
std_values = df[numerical_cols].std()

dispersion = pd.DataFrame({
    "Variance": variance_values,
    "Std Deviation": std_values
}).round(2)

print("Measures of Dispersion:")
display(dispersion)
```

```python
# Numerical columns
numerical_cols = ["age", "trestbps", "chol", "thalach", "oldpeak"]

# Plot a histogram (with KDE curve) for each numerical column
fig, axes = plt.subplots(2, 3, figsize=(16, 9))
axes = axes.flatten()

for i, col in enumerate(numerical_cols):
    sns.histplot(df[col], kde=True, bins=20, color="#00a8b5", ax=axes[i])
    axes[i].axvline(df[col].mean(), color="red", linestyle="--", label="Mean")
    axes[i].axvline(df[col].median(), color="green", linestyle="-", label="Median")
    axes[i].set_title(f"Distribution of {col}")
    axes[i].legend()

# Remove the extra empty plot
fig.delaxes(axes[5])

plt.tight_layout()
plt.show()
```

```python
# Skewness: measures the symmetry of the distribution
# Kurtosis: measures how heavy the tails are (scipy gives excess kurtosis, normal = 0)
shape_stats = pd.DataFrame({
    "Skewness": [skew(df[col]) for col in numerical_cols],
    "Kurtosis": [kurtosis(df[col]) for col in numerical_cols]
}, index=numerical_cols).round(3)

# Add a simple label for skewness
def skew_label(value):
    if value > 0.5:
        return "Right Skewed"
    elif value < -0.5:
        return "Left Skewed"
    else:
        return "Approximately Symmetric"

shape_stats["Skewness Type"] = shape_stats["Skewness"].apply(skew_label)

display(shape_stats)
```

## Reliability and Validity

| | Reliability | Validity |
|---|---|---|
| **Meaning** | Data is consistent and dependable | Data actually measures what it claims to measure |
| **Question** | Can we trust the data? | Is the data measuring the right thing? |
| **Example** | A scale shows 70 kg every time | A scale shows weight, not height |

### Reliability
Reliability means the data is **consistent and dependable**: repeating the measurement would give similar results.

- There are **no missing values**, so every variable is fully recorded.

### Validity
Validity means the data **actually measures what it claims to measure**.

- Variables like age, resting blood pressure (trestbps), cholesterol (chol), and maximum heart rate (thalach) are standard clinical indicators of heart health, so the dataset has good **content validity** for studying heart disease.
- All values fall within realistic medical ranges, with no impossible values (such as zero cholesterol).
- The `target` variable (heart disease or not) is a clear medical outcome, which supports validity.
- **Limitations:** the dataset has no lifestyle, genetic, or diet information, and the sample is small. So conclusions cannot be generalized to every population.

### Conclusion
After removing duplicates, the dataset is reasonably **reliable** and **valid** for statistical analysis, but the results should be interpreted with these limitations in mind.

### Population vs Sample

| | Population | Sample |
|---|---|---|
| **Meaning** | The complete group we want to study | A smaller part selected from the population |
| **Example here** | All patients in the cleaned dataset | A random group of patients taken from it |
| **Symbols** | Mean = μ, Std = σ, Size = N | Mean = x̄, Std = s, Size = n |

In real life, studying the whole population is often too costly or impossible, so we study a **sample** and use it to draw conclusions about the population. A good sample must be **random** and **representative**. In this project, the cleaned dataset is treated as the population, and we draw samples from it.

### What is Correlation?
Correlation measures the **strength and direction of the relationship** between two numerical variables. Its value is always between -1 and +1.

| Value | Meaning |
|---|---|
| Close to +1 | Strong positive relationship (when one increases, the other increases) |
| Close to -1 | Strong negative relationship (when one increases, the other decreases) |
| Close to 0 | No relationship |

- **Pearson correlation:** measures a straight-line (linear) relationship.
- **Spearman correlation:** measures a rank-based relationship, and works even when the data is not perfectly linear.

```python
# Numerical columns
numerical_cols = ["age", "trestbps", "chol", "thalach", "oldpeak"]

# Pearson correlation between all numerical columns
pearson_corr = df[numerical_cols].corr(method="pearson").round(2)
pearson_corr
```

```python
# Spearman correlation between all numerical columns
spearman_corr = df[numerical_cols].corr(method="spearman").round(2)
spearman_corr
```

```python
# Heatmap makes the correlation table easy to read
plt.figure(figsize=(8, 6))
sns.heatmap(pearson_corr, annot=True, cmap="coolwarm")
plt.title("Pearson Correlation Heatmap")
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.heatmap(spearman_corr, annot=True, cmap="coolwarm")
plt.title("Spearman Correlation Heatmap")
plt.show()
```

### Null and Alternative Hypothesis

- **Null Hypothesis (H0):** There is **no effect or no difference**. This is the default assumption.
- **Alternative Hypothesis (H1):** There **is** an effect or a difference.

**Decision rule (significance level = 0.05):**
- If **p-value < 0.05**, we **reject H0**. The result is statistically significant.
- If **p-value >= 0.05**, we **fail to reject H0**. There is not enough evidence of a difference.

| Test | Used for |
|---|---|
| T-Test | Compare the mean of a numerical variable between 2 groups |
| Chi-Square | Check the relationship between 2 categorical variables |
| ANOVA | Compare the mean of a numerical variable between 3 or more groups |

### Test 1: T-Test
**Question:** Is the average maximum heart rate (thalach) different between patients with heart disease and without?

- **H0:** The mean thalach is the same in both groups.
- **H1:** The mean thalach is different between the two groups.

```python
# Split the data into two groups
disease = df[df["target"] == 1]["thalach"]
no_disease = df[df["target"] == 0]["thalach"]

# Mean of each group
print("Mean thalach (heart disease)   :", disease.mean())
print("Mean thalach (no heart disease):", no_disease.mean())

# Perform the t-test
t_value, p_value = ttest_ind(disease, no_disease)

print("\nT-statistic:", t_value)
print("P-value    :", p_value)

# Decision
if p_value < 0.05:
    print("Result: Reject H0 (the difference is significant)")
else:
    print("Result: Fail to reject H0 (no significant difference)")
```

### Test 2: Chi-Square Test
**Question:** Is there a relationship between sex and heart disease?

- **H0:** Sex and heart disease are independent (no relationship).
- **H1:** Sex and heart disease are related.

```python
# Make a table that counts patients for each sex and target
table = pd.crosstab(df["sex"], df["target"])
print(table)

# Perform the chi-square test
chi2, p_value, dof, expected = chi2_contingency(table)

print("\nChi-square value:", chi2)
print("P-value         :", p_value)

# Decision
if p_value < 0.05:
    print("Result: Reject H0 (sex and heart disease are related)")
else:
    print("Result: Fail to reject H0 (no significant relationship)")
```

```python
# Make one group for each chest pain type
group0 = df[df["cp"] == 0]["thalach"]
group1 = df[df["cp"] == 1]["thalach"]
group2 = df[df["cp"] == 2]["thalach"]
group3 = df[df["cp"] == 3]["thalach"]

# Mean of each group
print("Mean thalach (cp = 0):", group0.mean())
print("Mean thalach (cp = 1):", group1.mean())
print("Mean thalach (cp = 2):", group2.mean())
print("Mean thalach (cp = 3):", group3.mean())

# Perform ANOVA
f_value, p_value = f_oneway(group0, group1, group2, group3)

print("\nF-statistic:", f_value)
print("P-value    :", p_value)

# Decision
if p_value < 0.05:
    print("Result: Reject H0 (at least one group is different)")
else:
    print("Result: Fail to reject H0 (no significant difference)")
```

### What is Normality Testing?
A **normal distribution** is the bell-shaped curve, where most values are near the mean and the data is symmetric. Many statistical tests (like the t-test and ANOVA) work best when the data is approximately normal, so we check this first.

The **Shapiro-Wilk test** checks whether a variable follows a normal distribution.

- **H0:** The data is normally distributed.
- **H1:** The data is not normally distributed.

**Decision rule (significance level = 0.05):**
- If **p-value > 0.05**, we **fail to reject H0**. The data looks normal.
- If **p-value <= 0.05**, we **reject H0**. The data is not normal.

```python
# Shapiro-Wilk test on cholesterol
stat, p_value = shapiro(df["chol"])

print("Statistic:", stat)
print("P-value  :", p_value)

# Decision
if p_value > 0.05:
    print("Result: Fail to reject H0 (data looks normal)")
else:
    print("Result: Reject H0 (data is not normal)")
```

```python
# Numerical columns
numerical_cols = ["age", "trestbps", "chol", "thalach", "oldpeak"]

# Run the test on every column
for col in numerical_cols:
    stat, p_value = shapiro(df[col])

    if p_value > 0.05:
        result = "Normal"
    else:
        result = "Not Normal"

    print(col, "| Statistic:", round(stat, 4), "| P-value:", round(p_value, 4), "|", result)
```

### What is a Confidence Interval?
A **confidence interval (CI)** is a range of values that is likely to contain the **true population mean**. Instead of giving only one number (the sample mean), it gives a range.

For example, a **95% confidence interval** of (240, 252) for cholesterol means we are 95% confident that the true average cholesterol of the population is between 240 and 252.

- A **narrow** interval means a more precise estimate.
- A **wide** interval means a less precise estimate.
- A larger sample size gives a narrower interval.

```python
# 95% confidence interval for the mean cholesterol
data = df["chol"]

mean = data.mean()          # sample mean
std_error = stats.sem(data) # standard error of the mean

# Confidence interval using the t-distribution
lower, upper = stats.t.interval(0.95, len(data) - 1, loc=mean, scale=std_error)

print("Mean of chol          :", round(mean, 2))
print("95% Confidence Interval:", round(lower, 2), "to", round(upper, 2))
```

```python
# Numerical columns
numerical_cols = ["age", "trestbps", "chol", "thalach", "oldpeak"]

# Calculate the 95% confidence interval for every column
for col in numerical_cols:
    data = df[col]
    mean = data.mean()
    std_error = stats.sem(data)
    lower, upper = stats.t.interval(0.95, len(data) - 1, loc=mean, scale=std_error)

    print(col, "| Mean:", round(mean, 2), "| 95% CI:", round(lower, 2), "to", round(upper, 2))
```

```python
# Plot the 95% confidence interval of cholesterol on the histogram
data = df["chol"]
mean = data.mean()
lower, upper = stats.t.interval(0.95, len(data) - 1, loc=mean, scale=stats.sem(data))

sns.histplot(data, kde=True)
plt.axvline(mean, color="red", label="Mean")
plt.axvline(lower, color="green", linestyle="--", label="Lower CI")
plt.axvline(upper, color="green", linestyle="--", label="Upper CI")
plt.title("Cholesterol with 95% Confidence Interval")
plt.legend()
plt.show()
```
