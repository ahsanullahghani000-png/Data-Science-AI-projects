```python
# Count missing values in each column
print(df.isnull().sum())
```

```python
# Fill missing values of chol with the median
df["chol"] = df["chol"].fillna(df["chol"].median())
```

```python
# How many rows are exact copies of another row?
print("Rows before:", len(df))
print("Duplicate rows:", df.duplicated().sum())
```

```python
df.duplicated().sum()
```

```python
# Remove the duplicate rows
df = df.drop_duplicates()

# Reset the row numbers (0, 1, 2, ...)
df = df.reset_index(drop=True)

print("Rows after:", len(df))
```

***Outlier Concept***

```python
# Numerical columns (outliers only make sense for these)
cols = ["age", "trestbps", "chol", "thalach", "oldpeak"]

print("Rows before:", len(df))

# Remove outliers from each column using the IQR method
for col in cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    df = df[(df[col] >= lower) & (df[col] <= upper)]

# Reset the row numbers
df = df.reset_index(drop=True)

print("Rows after :", len(df))
```

```python
# Boxplots after removing outliers
fig, axes = plt.subplots(1, 5, figsize=(16, 4))

for i, col in enumerate(cols):
    sns.boxplot(y=df[col], ax=axes[i])
    axes[i].set_title(col)

plt.tight_layout()
plt.show()
```

```python
# Columns that are categories (the numbers are just labels)
cat_cols = ["cp", "restecg", "slope", "thal"]

# Turn each category into its own 0/1 column
# drop_first=True removes one extra column per category (it is not needed)
X_encoded = pd.get_dummies(X, columns=cat_cols, drop_first=True, dtype=int)

print("Number of columns after:", X_encoded.shape[1])
X_encoded.head()
```

```python
# Numerical columns that need scaling
num_cols = ["age", "trestbps", "chol", "thalach", "oldpeak"]

# Make a copy of the data
X_standard = X_encoded.copy()

# Apply StandardScaler (mean becomes 0, std becomes 1)
scaler1 = StandardScaler()
X_standard[num_cols] = scaler1.fit_transform(X_standard[num_cols])

X_standard.head()
```

```python
# Make another copy of the data
X_minmax = X_encoded.copy()

# Apply MinMaxScaler (values become between 0 and 1)
scaler2 = MinMaxScaler()
X_minmax[num_cols] = scaler2.fit_transform(X_minmax[num_cols])

X_minmax.head()
```
