```python
# Columns that are categories (the numbers are just labels)
cat_cols = ["Department", "salary"]

# Turn each category into its own 0/1 column
# drop_first=True removes one extra column per category (it is not needed)
X_encoded = pd.get_dummies(X, columns=cat_cols, drop_first=True, dtype=int)

print("Number of columns after:", X_encoded.shape[1])
X_encoded.head()
```

```python
df['Department'].unique()
```

***Feature Scaling***

```python
# scale the numeric columns
num_cols = ["satisfaction_level", "last_evaluation", "number_project",
            "average_montly_hours", "time_spend_company"]

X_scaled = X_encoded.copy()
X_scaled[num_cols] = StandardScaler().fit_transform(X_scaled[num_cols])
X_scaled[num_cols].describe().round(2)
```

```python

# remove duplicates
df = df.drop_duplicates().reset_index(drop=True)
print("rows now:", len(df))

# X and y
y = df["left"]
X = df.drop("left", axis=1)

# encoding
X["salary"] = X["salary"].map({"low": 0, "medium": 1, "high": 2})
X_encoded = pd.get_dummies(X, columns=["Department"], drop_first=True, dtype=int)

# scaling
num_cols = ["satisfaction_level", "last_evaluation", "number_project",
            "average_montly_hours", "time_spend_company"]
X_scaled = X_encoded.copy()
X_scaled[num_cols] = StandardScaler().fit_transform(X_scaled[num_cols])

print("X_encoded shape:", X_encoded.shape)
X_scaled[num_cols].describe().round(2)
```

### Interpretation
- After removing duplicates, 11991 rows are left.
- After StandardScaler, every scaled column has a mean of 0 and a std of 1.
- time_spend_company has the largest maximum (4.99). Some employees stayed 10 years, far more than the average of about 3.4 years.
- time_spend_company counts whole years, so many employees share the same value and the 25% and 50% values are equal.

```python
# 80% for training, 20% for testing
# stratify=y keeps the same stay/left percentage in both parts
X_train, X_test, y_train, y_test = train_test_split(
    X_encoded, y, test_size=0.2, random_state=42, stratify=y)

print("X_train:", X_train.shape)
print("X_test :", X_test.shape)
```

```python
# percentage of employees who left, in each part
print("Train:")
print((y_train.value_counts(normalize=True) * 100).round(2))

print("\nTest:")
print((y_test.value_counts(normalize=True) * 100).round(2))
```

```python
num_cols = ["satisfaction_level", "last_evaluation", "number_project",
            "average_montly_hours", "time_spend_company"]

# copies for the models that need scaling (Logistic Regression, SVM, KNN, Naive Bayes)
X_train_scaled = X_train.copy()
X_test_scaled = X_test.copy()

scaler = StandardScaler()
X_train_scaled[num_cols] = scaler.fit_transform(X_train[num_cols])
X_test_scaled[num_cols] = scaler.transform(X_test[num_cols])
```

```python
# Logistic Regression
lr = LogisticRegression(max_iter=1000)
lr.fit(X_train_scaled, y_train)
lr_pred = lr.predict(X_test_scaled)
print("Accuracy:", accuracy_score(y_test, lr_pred))
```

```python
# Support Vector Machine
svm = SVC()
svm.fit(X_train_scaled, y_train)
svm_pred = svm.predict(X_test_scaled)
print("Accuracy:", accuracy_score(y_test, svm_pred))
```

```python
# K-Nearest Neighbors
knn = KNeighborsClassifier()
knn.fit(X_train_scaled, y_train)
knn_pred = knn.predict(X_test_scaled)
print("Accuracy:", accuracy_score(y_test, knn_pred))
```

```python
# Random Forest
rf = RandomForestClassifier(random_state=42)
rf.fit(X_train, y_train)
rf_pred = rf.predict(X_test)
print("Accuracy:", accuracy_score(y_test, rf_pred))
```

```python
# Decision Tree
dt = DecisionTreeClassifier(random_state=42)
dt.fit(X_train, y_train)
dt_pred = dt.predict(X_test)
print("Accuracy:", accuracy_score(y_test, dt_pred))
```

```python
# Naive Bayes
nb = GaussianNB()
nb.fit(X_train_scaled, y_train)
nb_pred = nb.predict(X_test_scaled)
print("Accuracy:", accuracy_score(y_test, nb_pred))
```

```python
# compare all six models in one table
names = ["Logistic Regression", "SVM", "KNN", "Decision Tree", "Random Forest", "Naive Bayes"]
preds = [lr_pred, svm_pred, knn_pred, dt_pred, rf_pred, nb_pred]

results = pd.DataFrame({
    "Model": names,
    "Accuracy": [accuracy_score(y_test, p) for p in preds],
    "Precision": [precision_score(y_test, p) for p in preds],
    "Recall": [recall_score(y_test, p) for p in preds],
    "F1-score": [f1_score(y_test, p) for p in preds]
})

results.round(3)
```
