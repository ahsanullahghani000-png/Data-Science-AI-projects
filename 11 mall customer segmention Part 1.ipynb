```python



pca_full = PCA(random_state=RANDOM_STATE)
pca_full.fit(X_scaled)

print("Explained Variance Ratio:", np.round(pca_full.explained_variance_ratio_, 3))
print("Cumulative Variance:    ", np.round(np.cumsum(pca_full.explained_variance_ratio_), 3))
```

```python
plt.figure(figsize=(8, 5))
plt.plot(range(1, len(pca_full.explained_variance_ratio_) + 1),
         np.cumsum(pca_full.explained_variance_ratio_),
         marker='o', color='#38bdf8')
plt.axhline(y=0.95, color='#fb7185', linestyle='--', label='95% Variance')
plt.xlabel('Number of Components')
plt.ylabel('Cumulative Explained Variance')
plt.title('PCA — Scree Plot')
plt.legend()
plt.show()
```

```python
pca = PCA(n_components=2, random_state=42)
X_pca = pca.fit_transform(X_scaled)

print("Total Variance Captured:", round(pca.explained_variance_ratio_.sum(), 3))
```

```python
plt.figure(figsize=(7, 5))
plt.scatter(X_pca[:, 0], X_pca[:, 1], c='#38bdf8', alpha=0.7, edgecolor='k', s=45)
plt.xlabel(f'PC1 ({pca.explained_variance_ratio_[0]*100:.1f}%)')
plt.ylabel(f'PC2 ({pca.explained_variance_ratio_[1]*100:.1f}%)')
plt.title('PCA — 2D Projection of Customers')
plt.grid(alpha=0.3)
plt.show()
```

```python
loadings = pd.DataFrame(
    pca.components_.T,
    columns=['PC1', 'PC2'],
    index=df.columns
)
loadings.round(3)
```

```python
plt.figure(figsize=(7, 5))
plt.scatter(X_tsne[:, 0], X_tsne[:, 1], c='#38bdf8', alpha=0.75, edgecolor='k', s=45)
plt.xlabel('t-SNE 1')
plt.ylabel('t-SNE 2')
plt.title('t-SNE — 2D Cluster Visualization')
plt.grid(alpha=0.3)
plt.show()
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout, Input

def build_model(hidden_layers, dropout_rate=0.2, lr=0.001):
    """Build and compile an MLP with the given hidden layer sizes."""
    model = Sequential()
    model.add(Input(shape=(X_train.shape[1],)))

    for i, neurons in enumerate(hidden_layers):
        model.add(Dense(neurons, activation='relu'))
        if dropout_rate > 0:
            model.add(Dropout(dropout_rate))

    model.add(Dense(1, activation='sigmoid'))

    model.compile(
        optimizer=tf.keras.optimizers.Adam(learning_rate=lr),
        loss='binary_crossentropy',
        metrics=['accuracy']
    )
    return model
```

```python
configs = {
    'Small (16)'        : [16],
    'Shallow (32)'      : [32],
    'Medium (32-16)'    : [32, 16],
    'Wide (128-64)'     : [128, 64],
    'Base (64-32-16)'   : [64, 32, 16],
    'Deep (64-32-16-8)' : [64, 32, 16, 8],
}

print(f"Testing {len(configs)} architectures")
for name, layers in configs.items():
    print(f"  {name}: {layers}")
```

```python
results = []

for name, layers in configs.items():
    print(f"\n Training: {name}")

    m = build_model(hidden_layers=layers, dropout_rate=0.2)

    history = m.fit(
        X_train, y_train,
        validation_split=0.2,
        epochs=15,              # short run for fair comparison
        batch_size=32,
        verbose=0
    )

    # Final metrics
    train_acc = history.history['accuracy'][-1]
    val_acc   = history.history['val_accuracy'][-1]
    val_loss  = history.history['val_loss'][-1]

    results.append({
        'Model'      : name,
        'Layers'     : str(layers),
        'Params'     : m.count_params(),
        'Train_Acc'  : round(train_acc, 4),
        'Val_Acc'    : round(val_acc, 4),
        'Val_Loss'   : round(val_loss, 4)
    })

    print(f"   Train Acc: {train_acc:.4f} | Val Acc: {val_acc:.4f} | Params: {m.count_params()}")
```

```python
results_df = pd.DataFrame(results).sort_values('Val_Acc', ascending=False)
results_df
```

```python
# Compare ReLU, Sigmoid, Tanh, and Softmax visually
x = np.linspace(-5, 5, 200)

relu    = np.maximum(0, x)
sigmoid = 1 / (1 + np.exp(-x))
tanh    = np.tanh(x)

fig, ax = plt.subplots(1, 3, figsize=(16, 4))

ax[0].plot(x, relu, color='#38bdf8', linewidth=2.5)
ax[0].axhline(0, color='gray', linewidth=0.6)
ax[0].axvline(0, color='gray', linewidth=0.6)
ax[0].set_title('ReLU: max(0, x)')
ax[0].grid(alpha=0.3)

ax[1].plot(x, sigmoid, color='#fb7185', linewidth=2.5)
ax[1].axhline(0, color='gray', linewidth=0.6)
ax[1].axvline(0, color='gray', linewidth=0.6)
ax[1].set_title('Sigmoid: 1/(1+e^-x)')
ax[1].grid(alpha=0.3)

ax[2].plot(x, tanh, color='#38bdf8', linewidth=2.5)
ax[2].axhline(0, color='gray', linewidth=0.6)
ax[2].axvline(0, color='gray', linewidth=0.6)
ax[2].set_title('Tanh: (e^x - e^-x)/(e^x + e^-x)')
ax[2].grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout, Input

def build_model(activation='relu', hidden_layers=[64, 32, 16], dropout_rate=0.2):
    """Build MLP with a specified activation function for hidden layers."""
    model = Sequential()
    model.add(Input(shape=(X_train.shape[1],)))

    for neurons in hidden_layers:
        model.add(Dense(neurons, activation=activation))
        model.add(Dropout(dropout_rate))

    model.add(Dense(1, activation='sigmoid'))   # binary output always sigmoid
    model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
    return model
```

```python
activations = ['relu', 'sigmoid', 'tanh']
act_results = []

for act in activations:
    print(f"\n▶ Training with activation: {act}")

    m = build_model(activation=act)

    hist = m.fit(
        X_train, y_train,
        validation_split=0.2,
        epochs=20,
        batch_size=32,
        verbose=0
    )

    act_results.append({
        'Activation' : act,
        'Train_Acc'  : round(hist.history['accuracy'][-1], 4),
        'Val_Acc'    : round(hist.history['val_accuracy'][-1], 4),
        'Val_Loss'   : round(hist.history['val_loss'][-1], 4),
        'Params'     : m.count_params()
    })

    print(f"   Val Acc: {act_results[-1]['Val_Acc']} | Val Loss: {act_results[-1]['Val_Loss']}")
```

```python
# For Softmax, we must use 2 output neurons + categorical_crossentropy
from tensorflow.keras.utils import to_categorical

y_train_cat = to_categorical(y_train, num_classes=2)
y_test_cat  = to_categorical(y_test,  num_classes=2)

def build_softmax_model(hidden_layers=[64, 32, 16], dropout_rate=0.2):
    model = Sequential()
    model.add(Input(shape=(X_train.shape[1],)))
    for neurons in hidden_layers:
        model.add(Dense(neurons, activation='relu'))
        model.add(Dropout(dropout_rate))
    model.add(Dense(2, activation='softmax'))    # 2 classes
    model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
    return model

softmax_model = build_softmax_model()
hist_soft = softmax_model.fit(
    X_train, y_train_cat,
    validation_split=0.2,
    epochs=20,
    batch_size=32,
    verbose=0
)

act_results.append({
    'Activation' : 'softmax (output)',
    'Train_Acc'  : round(hist_soft.history['accuracy'][-1], 4),
    'Val_Acc'    : round(hist_soft.history['val_accuracy'][-1], 4),
    'Val_Loss'   : round(hist_soft.history['val_loss'][-1], 4),
    'Params'     : softmax_model.count_params()
})
```

```python
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint

# Stop training when validation loss stops improving
early_stopping = EarlyStopping(
    monitor='val_loss',           # metric to watch
    patience=5,                   # wait 5 epochs before stopping
    restore_best_weights=True,    # restore weights from best epoch
    verbose=1
)

# Save the best model automatically
checkpoint = ModelCheckpoint(
    filepath='best_bank_model.keras',
    monitor='val_loss',
    save_best_only=True,
    verbose=1
)
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout, Input

# Use the best architecture identified in Step 03 (base 64-32-16)
model = Sequential([
    Input(shape=(X_train.shape[1],)),
    Dense(64, activation='relu'),
    Dropout(0.3),
    Dense(32, activation='relu'),
    Dropout(0.2),
    Dense(16, activation='relu'),
    Dense(1, activation='sigmoid')
])

model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
    loss='binary_crossentropy',
    metrics=['accuracy']
)

model.summary()
```

```python
history = model.fit(
    X_train, y_train,
    validation_split=0.2,        # 20% of training data used for validation
    epochs=100,                  # high number — early stopping will handle it
    batch_size=32,
    callbacks=[early_stopping, checkpoint],
    verbose=1
)
```

```python
fig, ax = plt.subplots(1, 2, figsize=(14, 5))

# ---- Accuracy curves ----
ax[0].plot(history.history['accuracy'],     label='Train',      color='#38bdf8', linewidth=2)
ax[0].plot(history.history['val_accuracy'], label='Validation', color='#fb7185', linewidth=2)
ax[0].set_title('Model Accuracy')
ax[0].set_xlabel('Epoch')
ax[0].set_ylabel('Accuracy')
ax[0].legend()
ax[0].grid(alpha=0.3)

# ---- Loss curves ----
ax[1].plot(history.history['loss'],     label='Train',      color='#38bdf8', linewidth=2)
ax[1].plot(history.history['val_loss'], label='Validation', color='#fb7185', linewidth=2)
ax[1].set_title('Model Loss')
ax[1].set_xlabel('Epoch')
ax[1].set_ylabel('Loss')
ax[1].legend()
ax[1].grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
stopped_epoch = len(history.history['loss'])
best_epoch    = int(np.argmin(history.history['val_loss'])) + 1
best_val_loss = min(history.history['val_loss'])
best_val_acc  = max(history.history['val_accuracy'])

print(f"Training stopped at epoch : {stopped_epoch}")
print(f"Best epoch (lowest val_loss): {best_epoch}")
print(f"Best validation loss      : {best_val_loss:.4f}")
print(f"Best validation accuracy  : {best_val_acc:.4f}")
```

```python
# Load the model that achieved the best val_loss during training
best_model = tf.keras.models.load_model('best_bank_model.keras')
print(" Best model loaded successfully")
best_model.summary()
```

```python
final_train_acc = history.history['accuracy'][-1]
final_val_acc   = history.history['val_accuracy'][-1]
gap = final_train_acc - final_val_acc

print(f"Final Train Acc : {final_train_acc:.4f}")
print(f"Final Val Acc   : {final_val_acc:.4f}")
print(f"Gap             : {gap:.4f}")

if gap > 0.05:
    print("  Overfitting detected — training acc much higher than validation")
elif final_train_acc < 0.75 and final_val_acc < 0.75:
    print("  Underfitting detected — both accuracies are low")
else:
    print(" Good fit — train and validation are close")
```
